```python
#Understanding the missing values and filling the missing values
data.isnull().sum()
```

```python
#we can fill the age column missing value with mean
data['Age']=data['Age'].fillna(data['Age'].mean(),inplace=True)
```

```python
data.isnull().sum()
```

```python
data['Embarked']=data['Embarked'].fillna('S')
```

```python
data.isnull().sum()
```

```python
data.drop(columns=['Cabin'],inplace=True)
```

```python
data.isnull().sum()
```

feature enginnering is the process of creating ,transforming or extracting important information from the features before making ML model ready.

```python
#feature engineering ---> Feature encoding (it is the process of converting categorical values into numerical)
#in above case we have sex,Pclass,Embarked

data.columns
```

```python
#get_dummies()

pclass = pd.get_dummies(data['Pclass'],drop_first=True,dtype=int)
pclass
```

```python
gender = pd.get_dummies(data['Sex'],dtype=int)
gender
```

```python
embark = pd.get_dummies(data['Embarked'],drop_first=True,dtype=int)
embark
```

```python
#concatenation of all tables

final_data = pd.concat([data,pclass,gender,embark],axis=1)
final_data
```

```python
final_data.columns
```

```python
final_data.drop(columns=['PassengerId', 'Pclass', 'Name','Ticket', 'Sex', 'Embarked'],inplace=True)
```
